# Regression Plots

Seaborn has many built-in capabilities for regression plots, however we won't really discuss regression until the machine learning section of the course, so we will only cover the **lmplot()** function for now.

**lmplot** allows you to display linear models, but it also conveniently allows you to split up those plots based off of features, as well as coloring the hue based off of features.

Let's explore how this works:

```python
import seaborn as sns
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

## lmplot()

sns.lmplot() is a Seaborn function used for plotting linear regression models with scatter plots. It's super handy when you want to visualize the relationship between two variables, including the regression line and confidence interval.

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex') # hue is the grouping variable
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='smoker') # hue is the grouping variable
```
